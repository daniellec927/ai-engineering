## Step 1: Import Libraries and load the environment variables

```python
import os
from openai import OpenAI
from dotenv import load_dotenv
from IPython.display import Markdown, display

load_dotenv()

OPENAI_API_KEY = os.getenv("OPENAI_API_KEY")

if OPENAI_API_KEY is None:
    raise Exception("API Key is missing.")
else:
    print(OPENAI_API_KEY[:8])

client = OpenAI()
```

## Step 2: Call the API

```python
response = client.chat.completions.create(
    model="gpt-4.1-mini",
    messages=[
        {"role": "system", "content": "You are a helpful assistant."},
        {"role": "user", "content": "I am traveling to Spain next week."}
    ]
)

reply = response.choices[0].message.content

display(Markdown(reply))
```

### In theory... (doing it manually for illustration)

```python
messages = [
    {"role": "system", "content": "You are a helpful assistant."},
    {"role": "user", "content": "I am traveling to Spain next week."},
    {"role": "assistant", "content": "That's exciting! Spain is a beautiful country with a rich culture and history. Do you have any specific cities or regions in mind that you'll be visiting?"},
    {"role": "user", "content": "Where am I traveling to next week?"}
] # history of chat we have to maintain to pass it to the llm

response = client.chat.completions.create(
    model="gpt-4.1-mini",
    messages=messages
)

reply = response.choices[0].message.content

display(Markdown(reply))
```
