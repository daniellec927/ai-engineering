### Import Libraries and load the environment variables

```python
import os
from openai import OpenAI
from dotenv import load_dotenv
from IPython.display import Markdown, display

load_dotenv()

OPENAI_API_KEY = os.getenv("OPENAI_API_KEY")

if OPENAI_API_KEY is None:
    raise Exception("API Key is missing.")
else:
    print(OPENAI_API_KEY[:8])

client = OpenAI()
```

### 1. Responses API: basic call

```python
response = client.responses.create(
    model="gpt-4.1-mini",
    instructions="You are a helpful assistant.",
    input="I am traveling to Spain next week.",
)

reply = response.output_text
display(Markdown(reply))
```

### 2. Responses API: conversation history

```python
response = client.responses.create(
    model="gpt-4.1-mini",
    instructions="You are a helpful assistant.",
    input="I am traveling to Spain next week.",
)

previous_id = response.id # storing ID for future use

reply = response.output_text
display(Markdown(reply))
```

```python
response = client.responses.create(
    model="gpt-4.1-mini",
    previous_response_id = previous_id,
    instructions="You are a helpful assistant.",
    input="Where am I traveling to next week?",
    # store=False # storing the response generated
)

previous_id = response.id

reply = response.output_text
display(Markdown(reply))
```

```python
response = client.responses.create(
    model="gpt-4.1-mini",
    previous_response_id = previous_id,
    instructions="You are a helpful assistant.",
    input="What is the capital?",
)

reply = response.output_text
display(Markdown(reply))
```

### 3. Responses API: Built-in Tools

```python
tone = "dramatic and engaging"

instructions = f"""
You are a helpful assistant that writes social media posts for LinkedIn, long-form. Use multiple emojis.
You will be supplied some instructions by the user, based on that input write a post.
Use the following tone: {tone}
"""

prompt = f"""
Write about the 4 top stories in AI this past week, according to reliable news media. Include dates.
"""

resp = client.responses.create(
    model="gpt-6-luna",
    instructions=instructions,
    input=prompt,
    tools=[{"type": "web_search"}]
)

display(Markdown(resp.output_text))
```
