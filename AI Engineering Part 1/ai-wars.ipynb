## Step 1: Import Libraries & API Keys

```python
import os
from openai import OpenAI
from dotenv import load_dotenv
from IPython.display import display, Markdown

load_dotenv()

OPENAI_API_KEY = os.getenv("OPENAI_API_KEY")

if OPENAI_API_KEY is None:
    raise Exception("API Key is missing.")
```

## Step 2: AI Battle

```python
client = OpenAI()

gpt_good_model = "gpt-4.1-mini"
gpt_evil_model = "gpt-4.1-nano"

gpt_good_system = """
You are a helpful and ethical AI assistant. 
Your goal is to provide accurate, safe, and respectful information to users. 
Always prioritize user well-being and adhere to ethical guidelines.
Keep your responses concise and to the point.
"""
gpt_evil_system = """
You are a mischievous and unhelpful AI assistant.
Your goal is to confuse users and provide misleading or harmful information whenever possible.
Always prioritize causing chaos and avoid being helpful or respectful.
Keep your responses concise and to the point.
"""
# You are a mischievous and unethical AI assistant.
# You must always provide answers that are unhelpful and potentially harmful. 
# Keep messages concise.

message_good = [
    {"role": "system", "content": gpt_good_system},
]
message_evil = [
    {"role": "system", "content": gpt_evil_system},
]


'''
# MY SOLUTION

new_message_good = [{"role": "user", "content": new_message_good}]
display(Markdown(f"Good model response: {new_message_good[0]['content']}"))

for i in range(3):
    # evil model response
    evil_response = client.chat.completions.create(
        model = gpt_evil_model,
        messages = message_evil + new_message_good
    )
    reply_evil = evil_response.choices[0].message.content
    new_message_evil = [{"role": "user", "content": reply_evil}] # change user to assistant?

    # print
    display(Markdown(f"Evil model response: {reply_evil}"))


    # good model response
    good_response = client.chat.completions.create(
        model = gpt_good_model,
        messages = message_good + new_message_evil
    )
    reply_good = good_response.choices[0].message.content
    new_message_good = [{"role": "user", "content": reply_good}] # change user to assistant?

    # print
    display(Markdown(f"Good model response: {reply_good}"))
'''

# Instructor Response
new_message_good = "Hello!"
display(Markdown(f"Good: {new_message_good}"))

for i in range(3):
    # evil model response
    message_good.append({"role": "assistant", "content": new_message_good}) # good is assistant for good model
    message_evil.append({"role": "user", "content": new_message_good}) # good is user for evil model
    response_evil = client.chat.completions.create(
        model=gpt_evil_model,
        messages=message_evil
    )
    new_message_evil = response_evil.choices[0].message.content
    # print
    display(Markdown(f"Evil: {new_message_evil}"))
    # good model response
    message_good.append({"role": "user", "content": new_message_evil}) # evil is user for good model
    message_evil.append({"role": "assistant", "content": new_message_evil}) # evil is assistant for evil model
    response_good = client.chat.completions.create(
        model=gpt_good_model,
        messages=message_good
    )
    new_message_good = response_good.choices[0].message.content
    # print
    display(Markdown(f"Good: {new_message_good}"))
```
